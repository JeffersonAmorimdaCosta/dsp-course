# Simulação

## Objetivo da simulação

A simulação tem como objetivo analisar o comportamento de senoides discretas e exponenciais complexas, verificando como a amplitude, a frequência e a fase influenciam a representação dos sinais.

Serão realizadas quatro análises:

1. Comparação de senoides com diferentes frequências;
2. Comparação de sinais com diferentes fases;
3. Comparação de sinais com diferentes amplitudes;
4. Representação das partes real e imaginária de uma exponencial complexa.

## Bibliotecas utilizadas

Para desenvolver a simulação, serão utilizadas as bibliotecas NumPy, Matplotlib e Seaborn.

* **NumPy:** criação das sequências e realização dos cálculos matemáticos;
* **Matplotlib:** construção dos gráficos;
* **Seaborn:** configuração visual dos gráficos.

## Implementação

Apresentar o código-fonte utilizado na simulação.

**Código da simulação:**

```python
# ============================================================
# IMPORTAÇÕES E CONFIGURAÇÕES
# ============================================================

import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(
    style="whitegrid",
    context="notebook"
)
```

```python
# ============================================================
# PARÂMETROS GERAIS
# ============================================================

n = np.arange(0, 41, 1)

# Frequências angulares em radianos por amostra
frequencias = [
    np.pi / 8,
    np.pi / 4,
    np.pi / 2
]

# Amplitude e fase iniciais
A = 1
phi = 0

# ============================================================
# 1. SENOides COM DIFERENTES FREQUÊNCIAS
# ============================================================

plt.figure(figsize=(12, 8))

for i, omega in enumerate(frequencias, 1):

    x = A * np.cos(omega * n + phi)

    plt.subplot(3, 1, i)

    plt.stem(n, x)

    plt.title(
        rf"Senoide discreta: $\omega_0={omega/np.pi:.2f}\pi$"
    )

    plt.xlabel("Índice da amostra (n)")
    plt.ylabel("Amplitude")

    plt.grid(True)

plt.tight_layout()
plt.show()

# ============================================================
# 2. ALTERAÇÃO DA FASE
# ============================================================

omega = np.pi / 4

fases = [
    0,
    np.pi / 4,
    np.pi / 2
]

plt.figure(figsize=(12, 5))

for fase in fases:

    x = A * np.cos(omega * n + fase)

    plt.plot(
        n,
        x,
        marker="o",
        label=rf"$\phi={fase/np.pi:.2f}\pi$"
    )

plt.title("Influência da fase na senoide discreta")
plt.xlabel("Índice da amostra (n)")
plt.ylabel("Amplitude")

plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()

# ============================================================
# 3. ALTERAÇÃO DA AMPLITUDE
# ============================================================

amplitudes = [
    0.5,
    1.0,
    2.0
]

plt.figure(figsize=(12, 5))

for amplitude in amplitudes:

    x = amplitude * np.cos(omega * n + phi)

    plt.plot(
        n,
        x,
        marker="o",
        label=f"A = {amplitude}"
    )

plt.title("Influência da amplitude na senoide discreta")
plt.xlabel("Índice da amostra (n)")
plt.ylabel("Amplitude")

plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()

# ============================================================
# 4. EXPONENCIAL COMPLEXA
# ============================================================

omega = np.pi / 4
A = 1
phi = 0

z = A * np.exp(1j * (omega * n + phi))

parte_real = np.real(z)
parte_imaginaria = np.imag(z)

plt.figure(figsize=(12, 8))

# Parte real
plt.subplot(2, 1, 1)

plt.stem(
    n,
    parte_real,
    linefmt="C0-",
    markerfmt="C0o",
    basefmt="k-"
)

plt.title("Parte real da exponencial complexa")
plt.xlabel("Índice da amostra (n)")
plt.ylabel("Parte real")

plt.grid(True)

# Parte imaginária
plt.subplot(2, 1, 2)

plt.stem(
    n,
    parte_imaginaria,
    linefmt="C1-",
    markerfmt="C1o",
    basefmt="k-"
)

plt.title("Parte imaginária da exponencial complexa")
plt.xlabel("Índice da amostra (n)")
plt.ylabel("Parte imaginária")

plt.grid(True)

plt.tight_layout()
plt.show()

# ============================================================
# 5. REPRESENTAÇÃO NO PLANO COMPLEXO
# ============================================================

plt.figure(figsize=(7, 7))

plt.plot(
    parte_real,
    parte_imaginaria,
    marker="o"
)

plt.axhline(0, color="black", linewidth=0.8)
plt.axvline(0, color="black", linewidth=0.8)

plt.title("Exponencial complexa no plano complexo")
plt.xlabel("Parte real")
plt.ylabel("Parte imaginária")

plt.axis("equal")
plt.grid(True)
plt.tight_layout()
plt.show()
```
