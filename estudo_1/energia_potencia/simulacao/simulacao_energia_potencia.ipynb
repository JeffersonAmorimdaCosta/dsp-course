# Simulação

## Metodologia

A simulação compara uma sequência de duração finita com uma sequência constante. Para a sequência finita, calcula-se diretamente a soma dos quadrados das amostras. Para a sequência constante, calcula-se a potência média em janelas de diferentes tamanhos, observando sua tendência.

## Implementação

Apresentar o código-fonte utilizado na simulação.

**Código da simulação:**

```python
# ============================================================  
# IMPORTAÇÕES
# ============================================================
import numpy as np  
import matplotlib.pyplot as plt  
import seaborn as sns

# ============================================================  
# CONFIGURAÇÕES
# ============================================================
sns.set_theme(style="whitegrid", context="notebook")
```

```python
# ============================================================
# SINAL DE ENERGIA
# ============================================================

n_energia = np.arange(-4, 5)

x_energia = np.array([
    0, 1, 2, 1, 0,
    -1, -2, -1, 0
])

E = np.sum(np.abs(x_energia) ** 2)
P_energia = np.mean(np.abs(x_energia) ** 2)

print("=" * 60)
print("SINAL DE ENERGIA")
print("=" * 60)
print(f"Energia total = {E:.4f}")
print(f"Potência média na janela = {P_energia:.4f}")


# ============================================================
# SINAL DE POTÊNCIA
# ============================================================

A = 2.0
N_values = np.arange(1, 101)
potencias = []

for N in N_values:

    n = np.arange(-N, N + 1)

    x_potencia = A * np.ones_like(n, dtype=float)

    P_N = np.mean(np.abs(x_potencia) ** 2)
    potencias.append(P_N)

P_teorica = A ** 2

print("\n")
print("=" * 60)
print("SINAL DE POTÊNCIA")
print("=" * 60)
print(f"Potência teórica = {P_teorica:.4f}")
print(f"Potência para N = 100 = {potencias[-1]:.4f}")


# ============================================================
# GRÁFICO DO SINAL DE ENERGIA
# ============================================================

plt.figure(figsize=(10, 4))

plt.stem(
    n_energia,
    x_energia,
    basefmt="k-"
)

plt.axhline(0, color="black", linewidth=0.8)

plt.title("Sinal de energia")
plt.xlabel("Índice n")
plt.ylabel("Amplitude")

plt.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()


# ============================================================
# GRÁFICO DO SINAL DE POTÊNCIA
# ============================================================

n_plot = np.arange(-10, 11)

x_plot = A * np.ones_like(n_plot, dtype=float)

plt.figure(figsize=(10, 4))

plt.stem(
    n_plot,
    x_plot,
    basefmt="k-"
)

plt.axhline(0, color="black", linewidth=0.8)

plt.title("Sinal de potência")
plt.xlabel("Índice n")
plt.ylabel("Amplitude")

plt.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()


# ============================================================
# CONVERGÊNCIA DA POTÊNCIA MÉDIA
# ============================================================

plt.figure(figsize=(10, 5))

sns.lineplot(
    x=N_values,
    y=potencias,
    linewidth=2,
    label="Potência média calculada"
)

plt.axhline(
    P_teorica,
    linestyle="--",
    linewidth=2,
    label="Potência teórica"
)

plt.title("Convergência da potência média")
plt.xlabel("N")
plt.ylabel("Potência média")

plt.legend()
plt.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()


# ============================================================
# RESUMO
# ============================================================

print("\n")
print("=" * 60)
print("RESUMO")
print("=" * 60)

print(f"Energia do sinal finito: {E:.4f}")
print(f"Potência média do sinal finito: {P_energia:.4f}")
print(f"Potência do sinal constante: {P_teorica:.4f}")
```

## Interpretação do código

* O primeiro sinal possui duração finita e é utilizado para calcular a energia total.  
* A energia é obtida somando o módulo ao quadrado de todas as amostras.  
* O segundo sinal possui amplitude constante e é utilizado para analisar a potência média.  
* A potência é calculada para diferentes valores de N.  
* O gráfico de convergência permite observar a tendência da potência média em direção ao valor teórico.  
* Os gráficos em haste facilitam a visualização das sequências discretas.
