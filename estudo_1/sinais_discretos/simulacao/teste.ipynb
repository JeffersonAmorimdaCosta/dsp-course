```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
x = np.linspace(0, 10, 100)
y1 = np.sin(x)
y2 = np.cos(x)

plt.plot(x, y1, label='Sine')
plt.plot(x, y2, label='Cosine')
plt.title('Sine and Cosine Functions')
plt.xlabel('x values')
plt.ylabel('Function values')
plt.legend()
plt.grid()
plt.show()
```
