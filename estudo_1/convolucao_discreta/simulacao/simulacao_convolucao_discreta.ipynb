# Simulação

O cálculo realizado manualmente pode ser verificado computacionalmente utilizando Python e a função `numpy.convolve()`.

A simulação será realizada em duas etapas. Primeiro, será reproduzido o exemplo original. Em seguida, a resposta ao impulso será modificada para observar como essa alteração influencia a saída.

## Simulação do Exemplo Original

As sequências utilizadas são:

$$
x[n]=\{1,2,1\}
$$

e:

$$
h[n]=\{1,1\}
$$

O código utiliza `numpy.convolve()` para realizar a convolução diretamente.

**Código da simulação:**

```python
# ============================================================
# IMPORTAÇÕES E CONFIGURAÇÕES
# ============================================================

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(
    style="whitegrid",
    context="paper",
    font="DejaVu Sans"
)
```

```python
# ============================================================
# DEFINIÇÃO DOS SINAIS
# ============================================================

x = np.array([1, 2, 1])
h = np.array([1, 1])


# ============================================================
# CÁLCULO DA CONVOLUÇÃO
# ============================================================

y = np.convolve(x, h)


# ============================================================
# ÍNDICES DOS SINAIS
# ============================================================

n_x = np.arange(len(x))
n_h = np.arange(len(h))
n_y = np.arange(len(y))


# ============================================================
# EXIBIÇÃO DOS RESULTADOS
# ============================================================

print("x[n] =", x)
print("h[n] =", h)
print("y[n] =", y)


# ============================================================
# TABELA DOS RESULTADOS
# ============================================================

resultado = pd.DataFrame({
    "n": n_y,
    "y[n]": y
})

print("\nResultado da convolução:")
print(resultado)


# ============================================================
# VISUALIZAÇÃO
# ============================================================

fig, axes = plt.subplots(3, 1, figsize=(10, 8))

axes[0].stem(n_x, x)
axes[0].set_title(r"Sinal de entrada $x[n]$")
axes[0].set_xlabel(r"$n$")
axes[0].set_ylabel(r"$x[n]$")
axes[0].set_xticks(n_x)

axes[1].stem(n_h, h)
axes[1].set_title(r"Resposta ao impulso $h[n]$")
axes[1].set_xlabel(r"$n$")
axes[1].set_ylabel(r"$h[n]$")
axes[1].set_xticks(n_h)

axes[2].stem(n_y, y)
axes[2].set_title(r"Saída $y[n] = x[n] * h[n]$")
axes[2].set_xlabel(r"$n$")
axes[2].set_ylabel(r"$y[n]$")
axes[2].set_xticks(n_y)

plt.tight_layout()
plt.show()
```

## Alteração da Resposta ao Impulso

Para investigar como \(h[n]\) modifica a saída, será utilizada uma segunda resposta ao impulso:

$$
h_2[n]=\{1,2\}
$$

A entrada permanece a mesma:

$$
x[n]=\{1,2,1\}
$$

A nova convolução será:

$$
y_2[n]=x[n]\ast h_2[n]
$$

**O cálculo pode ser realizado com:**

```python
# ============================================================
# SEGUNDO EXPERIMENTO
# ============================================================

h2 = np.array([1, 2])

y2 = np.convolve(x, h2)

n_h2 = np.arange(len(h2))
n_y2 = np.arange(len(y2))


# ============================================================
# EXIBIÇÃO DO SEGUNDO RESULTADO
# ============================================================

print("\nSegundo experimento:")
print("x[n]  =", x)
print("h2[n] =", h2)
print("y2[n] =", y2)


# ============================================================
# COMPARAÇÃO DAS SAÍDAS
# ============================================================

comparacao = pd.DataFrame({
    "n": n_y,
    "y[n] original": y,
    "y2[n] modificada": y2
})

print("\nComparação:")
print(comparacao)


# ============================================================
# VISUALIZAÇÃO DA COMPARAÇÃO
# ============================================================

plt.figure(figsize=(10, 5))

plt.stem(
    n_y,
    y,
    linefmt="C0-",
    markerfmt="C0o",
    basefmt="k-",
    label=r"$y[n]$ com $h[n]=\{1,1\}$"
)

plt.stem(
    n_y2,
    y2,
    linefmt="C1-",
    markerfmt="C1s",
    basefmt="k-",
    label=r"$y_2[n]$ com $h_2[n]=\{1,2\}$"
)

plt.title("Comparação das saídas")
plt.xlabel(r"$n$")
plt.ylabel("Amplitude")
plt.xticks(n_y)
plt.legend()

plt.tight_layout()
plt.show()
```
