# Simulação

## Metodologia

A simulação utiliza NumPy para manipulação das sequências, Matplotlib para os gráficos e Seaborn para configuração visual. Como os sinais são discretos, será utilizado o gráfico de haste (stem). Serão produzidos gráficos individuais e uma comparação conjunta.

## Implementação

Apresentar o código-fonte utilizado na simulação.

**Código da simulação:**

```python
# ============================================================
# IMPORTAÇÕES
# ============================================================
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# ============================================================
# CONFIGURAÇÕES
# ============================================================

sns.set_theme(style="whitegrid", context="notebook")

# Índices e sequência original
n = np.arange(-4, 5)
x = np.array([1, 2, 3, 2, 1, 0, -1, -2, -1])
```

```python
# ============================================================
# OPERAÇÕES
# ============================================================

# Original
n_original = n
x_original = x

# Atraso: x[n - 2]
n_atraso = n + 2
x_atraso = x

# Avanço: x[n + 2]
n_avanco = n - 2
x_avanco = x

# Inversão temporal: x[-n]
n_inversao = -n
x_inversao = x.copy()

# Ordenação para facilitar a visualização
ordem = np.argsort(n_inversao)
n_inversao = n_inversao[ordem]
x_inversao = x_inversao[ordem]

# Escalonamento: 2x[n]
n_amplificacao = n
x_amplificacao = 2 * x

# Inversão de amplitude: -x[n]
n_inversao_amp = n
x_inversao_amp = -x

# ============================================================
# FUNÇÃO PARA GRÁFICO INDIVIDUAL
# ============================================================

def plot_sinal(n_vals, x_vals, titulo):

    plt.figure(figsize=(10, 4))

    markerline, stemlines, baseline = plt.stem(
        n_vals,
        x_vals,
        basefmt="k-"
    )

    plt.setp(markerline, markersize=8)
    plt.setp(stemlines, linewidth=2)

    plt.axhline(0, color="black", linewidth=0.8)
    plt.axvline(0, color="gray", linestyle="--", linewidth=0.8)

    plt.title(titulo, fontsize=14)
    plt.xlabel("Índice n")
    plt.ylabel("Amplitude")

    plt.xticks(np.arange(-7, 8, 1))
    plt.grid(True, alpha=0.3)

    plt.tight_layout()
    plt.show()

# ============================================================
# GRÁFICOS INDIVIDUAIS
# ============================================================

plot_sinal(n_original, x_original, "Sinal original: x[n]")
plot_sinal(n_atraso, x_atraso, "Atraso de 2 amostras: x[n − 2]")
plot_sinal(n_avanco, x_avanco, "Avanço de 2 amostras: x[n + 2]")
plot_sinal(n_inversao, x_inversao, "Inversão temporal: x[−n]")
plot_sinal(n_amplificacao, x_amplificacao, "Escalonamento: 2x[n]")
plot_sinal(n_inversao_amp, x_inversao_amp, "Inversão de amplitude: −x[n]")

# ============================================================
# COMPARAÇÃO CONJUNTA
# ============================================================

operacoes = [
    ("Original: x[n]", n_original, x_original),
    ("Atraso: x[n − 2]", n_atraso, x_atraso),
    ("Avanço: x[n + 2]", n_avanco, x_avanco),
    ("Inversão temporal: x[−n]", n_inversao, x_inversao),
    ("Amplificação: 2x[n]", n_amplificacao, x_amplificacao),
    ("Inversão de amplitude: −x[n]", n_inversao_amp, x_inversao_amp)
]

fig, axes = plt.subplots(
    len(operacoes), 1,
    figsize=(12, 14),
    sharex=True
)

for ax, (titulo, n_vals, x_vals) in zip(axes, operacoes):

    markerline, stemlines, baseline = ax.stem(
        n_vals,
        x_vals,
        basefmt="k-"
    )

    plt.setp(markerline, markersize=7)
    plt.setp(stemlines, linewidth=2)

    ax.axhline(0, color="black", linewidth=0.8)
    ax.axvline(0, color="gray", linestyle="--", linewidth=0.8)

    ax.set_title(titulo, fontsize=12)
    ax.set_ylabel("Amplitude")
    ax.grid(True, alpha=0.25)

axes[-1].set_xlabel("Índice n")

fig.suptitle(
    "Comparação das operações realizadas sobre x[n]",
    fontsize=16,
    y=1.01
)

plt.tight_layout()
plt.show()
```

## Interpretação do código

- A sequência original é mantida como referência.
- No atraso, os índices são deslocados duas posições para a direita.
- No avanço, os índices são deslocados duas posições para a esquerda.
- Na inversão temporal, os índices são multiplicados por $-1$.
- Na amplificação, os valores são multiplicados por 2.
- Na inversão de amplitude, os valores são multiplicados por $-1$.
